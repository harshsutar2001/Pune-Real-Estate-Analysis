```python
df['total rooms']=df['num_bedrooms']+df['num_bathrooms']
```

```python
df
```

```python
df['luxury_index']=np.where(df['price_per_sqrt']>3000,'high','medium')
```

```python
df
```

```python
df['swimming_ful'] = df['swimming_ful'].map({'Yes': True, 'No': False})
df['garden'] = df['garden'].map({'Yes': True, 'No': False})
df['has_pool_or_garden'] = df['swimming_ful'] | df['garden']

```

```python
df
```

```python
df['price_category'] = pd.cut(df['price'],
                              bins=[0, 2000000, 5000000, df['price'].max()],
                              labels=['Low','Medium','High'])

df
```

# Distrubution of price


```python
plt.figure(figsize=(8,5))
sns.histplot(df['price'], bins=30, kde=True,color='red')
plt.title("Price Distribution")
plt.show()
```

# b) Price vs Square Feet

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='square_feet', y='price', hue='luxury_index', data=df)
plt.title("Price vs Square Feet")
plt.show()
```

# c) Average Price by Area

```python
plt.figure(figsize=(10,5))
sns.barplot(x='area', y='price', data=df.groupby('area')['price'].mean().reset_index(),color='yellow')
plt.xticks(rotation=45)
plt.title("Average Price by Area")
plt.show()
```

# d) Correlation Heatmap

```python

numeric_df = df.select_dtypes(include=np.number)


plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation between Numeric Features")
plt.show()

```

# Categorical Analysis

```python
# Count of properties by luxury index
sns.countplot(x='luxury_index', data=df,color='green')
plt.show()

# Count of properties with pool or garden
sns.countplot(x='has_pool_or_garden', data=df,color='blue')
plt.show()
```
